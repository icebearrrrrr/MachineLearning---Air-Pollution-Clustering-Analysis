# Data Cleaning

1. Handle the columns that are useful accordingly.
2. Invalid Data: Validate formats (especially for dates and times).
3. Missing Data: Identify and handle null values or missing data.
4. Contaminated Data: Look for any unrealistic or incorrect data.
5. Outliers: Detect and address outliers.
6. Duplicate Data: Identify and remove duplicates.

## Loading the dataset
the csv file is seperated differently than normal, with ';' being the seperator for each column and ',' being the decimal point for each data point.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#load data from local drive.
df=pd.read_csv('AirQuality.csv',sep=";",decimal=',')
df.info()
df.head()
```

```python
#safe it in proper format,(',' as seperator and '.' as decimal)
df.to_csv('AirQuality_processed.csv', index=False)
```

## 1. Handle columns
dropping the last two columns since have no values. This can be clearly seen since there is no name for the columns and there is no non-null values.

```python
df = df.drop(df.columns[-2:], axis=1)
df.head()
```

## 2. Invalid data
Making the date-time format join together following actual datatype(datetime) and removing the columns date and time since combined into 1 column already.

```python
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H.%M.%S')
df = df.drop(columns=['Date', 'Time'])  # Drop the original Date and Time columns if not needed
df.insert(0, 'Datetime', df.pop('Datetime'))  # Moves 'Datetime' to the first column
df.info()
```

```python
print(df.shape)
```

```python
df.describe()
```

## 3. Missing Data
Deal with all null values and missing data appropriately.

```python
print(df.isnull().sum())
df[df.isnull().any(axis=1)]  # Displays rows with at least one missing value
```

since the last 114 rows have all the columns with null values, the best way to handle this data is by removing the rows entirelyAs the null values brings nothing to the clustering, these rows are dropped.

```python
df = df.dropna(how='all')  # Removes rows where all columns are NaN
print(df.isnull().sum())
print(df.shape)
```

## 4. Contaminated Data
Identifying the values that does not make any sense and deal with them. There are negative values for all of the variables, even though it makes no sense. In the dataset description from kaggle, it says missing values are tagged with -200 value. So we will deal with them accordingly.

```python
#replace all -200 values with nan
df.replace(-200, np.nan, inplace=True)
missing_data = df.isnull().sum()
missing_data_percentage = (missing_data/len(df)) * 100
print(missing_data_percentage)
```

Dropping the NMHC(GT) column because there is too many missing values for it to be a valuable column

```python
df = df.drop(columns= ['NMHC(GT)'])
```

```python
df[df['RH'].isnull()]
```

All records/rows where RH is null will be removed, this is because it seems that all columns that has 366 null values are connected, thus most of the values are NaN, making it unsuable for analysis. This can be done because even after removing these rows, there are still enough records for clustering. The cause of the null value is likely due to either sensor failure or failure to record the actual values

```python
df = df.dropna(subset=['RH'])
missing_data = df.isnull().sum()
missing_data_percentage = (missing_data/len(df)) * 100
print(missing_data_percentage)
print(df.shape)
```

Since the missing values is still considered a rather large percentage of the records(around 18%), we will remove all the records where NO2(GT) is null, then the remaining null values will be filled with the mean of the respective columns

```python
df = df.dropna(subset=['NO2(GT)'])
missing_data = df.isnull().sum()
missing_data_percentage = (missing_data/len(df)) * 100
print(missing_data_percentage)
print(df.shape)
```

All remaining null values in the "CO(GT)" column is replaced with its mean value.

```python
df['CO(GT)'] = df['CO(GT)'].fillna(df['CO(GT)'].mean())
print(df.isnull().sum())
print(df.shape)
```

Outliers are calculated using the z score calculation with a threshhold of 3. These outliers are then removed, this is to avoid the outliers effecting our clusters later on.

```python
# # Convert to datetime and set index (already done)
df['Datetime'] = pd.to_datetime(df['Datetime'])  
df.set_index('Datetime', inplace=True)  

# Select numerical columns
numerical_features = df.select_dtypes(include=[np.number]).columns

# Compute Z-scores
z_scores = df[numerical_features].apply(zscore)

# Define threshold
threshold = 3
outliers = (np.abs(z_scores) > threshold)

# Filter out rows with any outlier in numerical columns
df = df[~outliers.any(axis=1)]

# Optional: Reset index if needed
df.reset_index(inplace=True)

print(f"Shape after removing outlier: {df.shape}")
```

## 3. Average values according to day of the week

```python
# Ensure datetime format
df_cleaned['Datetime'] = pd.to_datetime(df_cleaned['Datetime'])  
df_cleaned.set_index('Datetime', inplace=True)  

# Create a new column for the day of the week
df_cleaned['Day_of_Week'] = df_cleaned.index.day_name()

# Group by day of the week and compute the mean
df_weekly_avg = df_cleaned.groupby('Day_of_Week').mean()

# Reorder to ensure correct order of days
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df_weekly_avg = df_weekly_avg.reindex(day_order)

# Select numerical features
numerical_features = df_cleaned.select_dtypes(include=[np.number]).columns

# Plot bar graphs for each numerical feature
for feature in numerical_features:
    plt.figure(figsize=(10, 5))
    plt.bar(df_weekly_avg.index, df_weekly_avg[feature], color='skyblue', alpha=0.8)
    plt.xlabel("Day of the Week")
    plt.ylabel("Average Value")
    plt.title(f"Average {feature} for Each Day of the Week")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Reset index if needed
df_cleaned.reset_index(inplace=True)
df_cleaned = df_cleaned.drop(columns= ['Day_of_Week'])
```

The values appear lower on the weekends, especially Sunday, this may be due to the fact that people find Sunday as a resting day and thus stay at home more often, reducing carbon emisions.

## 4. Average values for hours in a day

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ensure datetime format
df_cleaned['Datetime'] = pd.to_datetime(df_cleaned['Datetime'], errors='coerce')  # Coerce fixes invalid date errors
df_cleaned.set_index('Datetime', inplace=True)


# Create a new column for the hour of the day
df_cleaned['Hour_of_Day'] = df_cleaned.index.hour

# Group by hour of the day and compute the mean
df_hourly_avg = df_cleaned.groupby('Hour_of_Day').mean()

# Select numerical features
df_hourly_avg.reset_index(inplace=True)

# Plot individual line graphs for each numerical feature
for feature in numerical_features:
    plt.figure(figsize=(10, 5))
    plt.plot(df_hourly_avg['Hour_of_Day'], df_hourly_avg[feature], marker='o', linestyle='-', color='royalblue', alpha=0.8)
    plt.xlabel("Hour of the Day")
    plt.ylabel("Average Value")
    plt.title(f"Average {feature} for Each Hour of the Day")
    plt.xticks(range(0, 24))  # Ensure x-axis has all 24 hours
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()


# Reset index if needed
df_cleaned.reset_index(inplace=True)
df_cleaned = df_cleaned.drop(columns= ['Hour_of_Day'])
```

Most of these results are to be expected, high peaks in gas emisions during 7-9am(time people go to office) and 6-8pm(time people get off work).
