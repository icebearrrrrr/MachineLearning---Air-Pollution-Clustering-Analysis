```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.cluster import MeanShift
import hdbscan
from sklearn.cluster import Birch
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.cluster import SpectralClustering
from sklearn.cluster import OPTICS
from tqdm import tqdm


df_cleaned = pd.read_csv('cleaned.csv')
df_cleaned['Datetime'] = pd.to_datetime(df_cleaned['Datetime'], errors='coerce')
df_cleaned.dtypes
```

## 1. OPTICS(Base)

```python
# Drop non-numeric column
df_numeric = df_cleaned.drop(columns=['Datetime'])

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_numeric)


# Elbow Method to determine optimal K
wcss = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(8,5))
sns.set_style("whitegrid")
plt.plot(K_range, wcss, marker='o', linestyle='-', color='b')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.title('Elbow Method for Optimal K')
plt.show()
```

```python
optics = OPTICS(metric="manhattan" )

# Reduce dimensions to 2D using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Fit OPTICS
df_cleaned['Cluster_OPTICS'] = optics.fit_predict(X_pca)

# Check the number of data points in each cluster
print(df_cleaned['Cluster_OPTICS'].value_counts())

sil_score_optics = silhouette_score(X_pca, df_cleaned['Cluster_OPTICS'])
print(f'Silhouette Score for OPTICS: {sil_score_optics:.3f}')


# Plot the clusters
plt.figure(figsize=(8, 5))
# Note: OPTICS can return -1 for noise points, using a colormap that handles this well
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=df_cleaned['Cluster_OPTICS'], palette='tab10', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Clusters Visualization with OPTICS (PCA 2D)')
plt.legend(title='Cluster')
plt.show()

# Calculate cluster summary statistics (excluding noise points if any)
cluster_summary_optics = df_cleaned[df_cleaned['Cluster_OPTICS'] >= 0].groupby('Cluster_OPTICS').mean()
```

## OPTICS(Tuned)

min_sample: increasing it for Larger, fewer clusters: Small, sparse clusters (and noise) will tend to get merged into larger ones or marked as noise.

xi: Fewer, larger clusters: Subtler density variations are ignored and absorbed into existing clusters

min_cluster_size: avoid smaller clusters

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.OPTICS.html

```python
# === 1. Define parameter grid at the top ===
param_grid = {
    'min_samples': [10,15,20],
    'xi': [0.005,0.01,0.05],
    'min_cluster_size': [0.05],
}

# === 3. Initialize search trackers ===
best_silhouette = -1
best_params = {}
results = []

total_combinations = (
    len(param_grid['min_samples']) *
    len(param_grid['xi']) *
    len(param_grid['min_cluster_size'])
)
print(f"Testing {total_combinations} parameter combinations...")

# === 4. Grid search with nested loops ===
for min_samples in tqdm(param_grid['min_samples'], desc="min_samples"):
    for xi in param_grid['xi']:
        for min_cluster_size in param_grid['min_cluster_size']:
            try:
                # Build and fit OPTICS
                optics = OPTICS(
                    min_samples=min_samples,
                    xi=xi,
                    min_cluster_size=min_cluster_size
                )
                clusters = optics.fit_predict(X_pca)

                # Count clusters (excluding noise)
                n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)

                # Only compute silhouette when >1 cluster and enough samples
                if 1 < n_clusters < len(X_pca) - 1:
                    mask = clusters != -1
                    if mask.sum() > n_clusters:
                        sil_score = silhouette_score(X_pca[mask], clusters[mask])

                        # Record this run
                        results.append({
                            'min_samples': min_samples,
                            'xi': xi,
                            'min_cluster_size': min_cluster_size,
                            'n_clusters': n_clusters,
                            'silhouette': sil_score,
                            'noise_points': int((clusters == -1).sum())
                        })

                        # Update best if improved
                        if sil_score > best_silhouette:
                            best_silhouette = sil_score
                            best_params = {
                                'min_samples': min_samples,
                                'xi': xi,
                                'min_cluster_size': min_cluster_size
                            }

            except Exception as e:
                print(f"Error with params min_samples={min_samples}, xi={xi}, "
                      f"min_cluster_size={min_cluster_size}: {e}")

# === 5. Summarize and visualize results ===
results_df = pd.DataFrame(results)

if not results_df.empty:
    print("\nTop 5 parameter combinations by silhouette score:")
    print(results_df.sort_values('silhouette', ascending=False).head(), "\n")

    print(f"Best parameters: {best_params}")
    print(f"Best silhouette score: {best_silhouette:.3f}\n")

    # Fit final model with best params
    best_optics = OPTICS(**best_params)
    df_cleaned['Cluster_OPTICS_Tuned'] = best_optics.fit_predict(X_pca)

    # 5a. Cluster scatter plot
    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        x=X_pca[:, 0], y=X_pca[:, 1],
        hue=df_cleaned['Cluster_OPTICS_Tuned'],
        palette='tab10', alpha=0.7
    )
    plt.xlabel('PC 1')
    plt.ylabel('PC 2')
    plt.title(f'OPTICS Clustering (silhouette={best_silhouette:.3f})')
    plt.legend(title='Cluster')
    plt.show()

    # 5b. Cluster summary stats
    cluster_summary_optics_tuned = df_cleaned[df_cleaned['Cluster_OPTICS_Tuned'] >= 0] \
        .groupby('Cluster_OPTICS_Tuned') \
        .mean()
else:
    print("No valid parameter combinations were found.")
```

## Gaussian Mixture(Tuned)

n_components_range = decide the amount of clusters, try a multitude of range to see which is the best.
covariance_types String describing the type of covariance parameters to use. Must be one of:

                ‘full’: each component has its own general covariance matrix.

                ‘tied’: all components share the same general covariance matrix.

                ‘diag’: each component has its own diagonal covariance matrix.

                ‘spherical’: each component has its own single variance
                
https://scikit-learn.org/stable/modules/generated/sklearn.mixture.GaussianMixture.html

```python

# Range of components to try
n_components_range = range(2, 11)
covariance_types = ['full', 'tied', 'diag', 'spherical']

# Store results
silhouette_scores = []
models = []

# Try different combinations
for n_components in n_components_range:
    for cov_type in covariance_types:
        gmm = GaussianMixture(
            n_components=n_components,
            covariance_type=cov_type,
            random_state=42,
            n_init=10  # Multiple initializations
        )
        
        gmm.fit(X_pca)
        labels = gmm.predict(X_pca)
        
        # Silhouette score (only if n_components > 1)
        if n_components > 1:
            sil_score = silhouette_score(X_pca, labels)
        else:
            sil_score = 0
            
        # Store results
        models.append((n_components, cov_type))
        silhouette_scores.append(sil_score)
        
        print(f"Components: {n_components}, Covariance: {cov_type}, Silhouette: {sil_score:.3f}")

best_silhouette_idx = np.argmax(silhouette_scores)

print(f"Best model by Silhouette: {models[best_silhouette_idx]}")

# Apply the best model
best_n_components, best_cov_type = models[best_silhouette_idx]  # You can choose which metric to use
best_gmm = GaussianMixture(
    n_components=best_n_components,
    covariance_type=best_cov_type,
    random_state=42,
    n_init=10
)

df_cleaned['Cluster_GMM_Tuned'] = best_gmm.fit_predict(X_pca)

print(df_cleaned['Cluster_GMM_Tuned'].value_counts())
# Compute silhouette score
sil_score_gmm_tuned = silhouette_score(X_pca, df_cleaned['Cluster_GMM_Tuned'])
print(f'Silhouette Score for GMM: {sil_score_gmm_tuned:.3f}')

# Visualize clusters (if PCA was applied)
plt.figure(figsize=(8, 5))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=df_cleaned['Cluster_GMM_Tuned'], palette='tab10', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Gaussian Mixture Clustering (PCA 2D)')
plt.legend(title='Cluster')
plt.show()

# Summary of clusters
cluster_summary_gmm_tuned = df_cleaned.groupby('Cluster_GMM_Tuned').mean()
```

## HDBSCAN(Tuned)

min_cluster_size: Higher value to get bigger clusters, effeectively reducing the number of cluster

min_samples:The higher the value, the fewer, denser clusters: You’ll see a conservative clustering that focuses on the densest regions.

metrics: The metric to use when calculating distance between instances in a feature array.

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.HDBSCAN.html

```python
best_score = -1
best_params = None

# Define search grid
min_cluster_sizes = [150, 200, 250, 300]
min_samples_list = [None, 25, 50, 100, 150]   # None → defaults to min_cluster_size
metrics = ['euclidean', 'manhattan']

for mcs in min_cluster_sizes:
    for ms in min_samples_list:
        for metric in metrics:
            # Instantiate HDBSCAN with chosen params
            hdb = hdbscan.HDBSCAN(
                min_cluster_size=mcs,
                min_samples=ms or mcs,
                metric=metric,
            )
            labels = hdb.fit_predict(X_pca)
            
            # Exclude noise for silhouette calculation
            mask = labels != -1
            unique_clusters = set(labels[mask])
            
            if len(unique_clusters) > 1:
                score = silhouette_score(X_pca[mask], labels[mask], metric=metric)
                print(f"mcs={mcs:<2}, ms={str(ms):<4}, metric={metric:<9} → silhouette={score:.3f}")
                
                if score > best_score:
                    best_score = score
                    best_params = (mcs, ms, metric)
            else:
                print(f"mcs={mcs:<2}, ms={str(ms):<4}, metric={metric:<9} → silhouette=NA")

print("\nBest silhouette score: "
      f"{best_score:.3f} with min_cluster_size={best_params[0]}, "
      f"min_samples={best_params[1]}, metric='{best_params[2]}'")

# FIX: Correctly unpack the parameters
hdb = hdbscan.HDBSCAN(
    min_cluster_size=best_params[0],
    min_samples=best_params[1] or best_params[0],  # Handle None case
    metric=best_params[2],
)

df_cleaned['Cluster_HDBSCAN_Tuned'] = hdb.fit_predict(X_pca)
labels = df_cleaned['Cluster_HDBSCAN_Tuned']

# HDBSCAN uses -1 for noise
core_labels = labels[labels != -1]
if len(set(core_labels)) > 1:
    sil_score_hdbscan = silhouette_score(
        X_pca[labels != -1],
        core_labels
    )
    print(f'Silhouette Score for HDBSCAN (excluding noise): {sil_score_hdbscan:.3f}')
else:
    print('Silhouette Score for HDBSCAN: Not applicable (fewer than 2 clusters detected)')

# -------------------------------
# 3. Cluster size counts
# -------------------------------
print(df_cleaned['Cluster_HDBSCAN_Tuned'].value_counts().sort_index())

# -------------------------------
# 4. Visualize clusters (PCA 2D)
# -------------------------------
plt.figure(figsize=(8, 5))
unique_labels = set(labels)
n_clusters = len(unique_labels - {-1})
palette = sns.color_palette('tab10', n_colors=n_clusters)
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=labels,
    palette={label: palette[i]
             for i, label in enumerate(sorted(unique_labels - {-1}))}
            | {-1: (0.5, 0.5, 0.5)},
    alpha=0.7,
    legend='full'
)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('HDBSCAN Clustering (PCA 2D)')
plt.legend(title='Cluster', loc='best')
plt.show()

# -------------------------------
# 5. Summary of clusters
# -------------------------------
cluster_summary_hdbscan_tuned = df_cleaned.groupby('Cluster_HDBSCAN_Tuned').mean()
```

## Mean-Shift(Tuned)

Bandwidth used in the flat kernel. Mean Shift is a mode-seeking clustering algorithm. For each data point, it places a kernel (often a Gaussian or flat “ball”) of radius = bandwidth around the point and computes the mean of all points inside that kernel. It then “shifts” the kernel to that mean and repeats until convergence—this converged position is a density mode. Points whose kernels converge to the same mode are grouped into the same cluster. 
A larger bandwidth will make for larger radius and larger clusters.

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.MeanShift.html

```python
# 1. Define a grid of bandwidths to try
bandwidths = np.linspace(0.2, 2.0, 10)  # e.g. from 0.2 to 2.0 in 10 steps

best_score = -1
best_bw = None
results = []

# 2. Loop over bandwidths
for bw in tqdm(bandwidths, desc="Bandwidth"):
    ms = MeanShift(bandwidth=bw)
    labels = ms.fit_predict(X_pca)         # or X_scaled if you prefer
    n_clusters = len(np.unique(labels))
    
    # Only compute silhouette if we have >1 cluster
    if n_clusters > 1:
        score = silhouette_score(X_pca, labels)
    else:
        score = -1.0
    
    results.append({'bandwidth': bw,
                    'n_clusters': n_clusters,
                    'silhouette': score})
    
    if score > best_score:
        best_score = score
        best_bw = bw

# 3. Summarize results
results_df = pd.DataFrame(results)
print("Top 5 bandwidths by silhouette:")
print(results_df.sort_values('silhouette', ascending=False).head(), "\n")
print(f"Best bandwidth: {best_bw:.3f}  →  Silhouette: {best_score:.3f}")

# 4. Refit with best bandwidth and visualize
best_ms = MeanShift(bandwidth=best_bw)
df_cleaned['Cluster_MS_Tuned'] = best_ms.fit_predict(X_pca)
print(df_cleaned['Cluster_MS_Tuned'].value_counts())

plt.figure(figsize=(8, 5))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1],
                hue=df_cleaned['Cluster_MS_Tuned'],
                palette='tab10', alpha=0.7)
plt.xlabel('PC 1')
plt.ylabel('PC 2')
plt.title(f'Mean Shift (bw={best_bw:.2f}, silhouette={best_score:.3f})')
plt.legend(title='Cluster')
plt.show()

# 5. Cluster summary
cluster_summary_ms_tuned = df_cleaned[df_cleaned['Cluster_MS_Tuned'] >= 0] \
                       .groupby('Cluster_MS_Tuned').mean()
```

## SPECTRAL(Tuned)

n_clusters: Number of clusters

affinity: nearest_neightbours:Connect each point only to its k-nearest neighbors.
          rbf: Fully connect the graph, weighting edges by an RBF (Gaussian) kernel

n_neighbours: Large (e.g. 20): Broader connectivity; incorporates more distant points into each node’s neighborhood. Graph becomes denser, which can smooth over small gaps and yield fewer, larger clusters.

gamma: Large: Kernel decays quickly—only very close points have high similarity. Graph is sparse in practice; clusters focus on tight, high‑density pockets

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.SpectralClustering.html

```python
# Define hyperparameter grid
param_grid = {
    'n_clusters': range(2, 7),               # Test 2 to 6 clusters
    'affinity': ['nearest_neighbors', 'rbf'],
    'n_neighbors': [5, 10, 15, 20],          # Only relevant for affinity='nearest_neighbors'
    'gamma': [0.1, 1.0, 10.0]                # Only relevant for affinity='rbf'
}

# Create lists to store results
results = []

# Iterate through hyperparameters
print("Running hyperparameter tuning for Spectral Clustering...")
for n_clusters in tqdm(param_grid['n_clusters']):
    for affinity in param_grid['affinity']:
        if affinity == 'nearest_neighbors':
            for n_neighbors in param_grid['n_neighbors']:
                # Create model with current hyperparameters
                spectral = SpectralClustering(
                    n_clusters=n_clusters,
                    affinity=affinity,
                    n_neighbors=n_neighbors,
                    random_state=42
                )
                
                # Fit model and predict clusters
                cluster_labels = spectral.fit_predict(X_pca)
                
                # Compute silhouette score if there's more than one cluster
                if len(set(cluster_labels)) > 1:
                    sil_score = silhouette_score(X_pca, cluster_labels)
                else:
                    sil_score = float('-inf')  # Assign negative infinity if only one cluster
                
                # Save results
                results.append({
                    'n_clusters': n_clusters,
                    'affinity': affinity,
                    'n_neighbors': n_neighbors,
                    'gamma': 'N/A',
                    'silhouette_score': sil_score,
                    'cluster_labels': cluster_labels
                })
                
        elif affinity == 'rbf':
            for gamma in param_grid['gamma']:
                # Create model with current hyperparameters
                spectral = SpectralClustering(
                    n_clusters=n_clusters,
                    affinity=affinity,
                    gamma=gamma,
                    random_state=42
                )
                
                # Fit model and predict clusters
                cluster_labels = spectral.fit_predict(X_pca)
                
                # Compute silhouette score if there's more than one cluster
                if len(set(cluster_labels)) > 1:
                    sil_score = silhouette_score(X_pca, cluster_labels)
                else:
                    sil_score = float('-inf')  # Assign negative infinity if only one cluster
                
                # Save results
                results.append({
                    'n_clusters': n_clusters,
                    'affinity': affinity,
                    'n_neighbors': 'N/A',
                    'gamma': gamma,
                    'silhouette_score': sil_score,
                    'cluster_labels': cluster_labels
                })

# Convert results to DataFrame and sort by silhouette score
results_df = pd.DataFrame(results)
results_df = results_df.sort_values('silhouette_score', ascending=False)

# Print the top 5 parameter combinations
print("\nTop 5 parameter combinations:")
print(results_df[['n_clusters', 'affinity', 'n_neighbors', 'gamma', 'silhouette_score']].head())

# Get the best parameters
best_params = results_df.iloc[0]
print("\nBest parameters:")
print(f"n_clusters: {best_params['n_clusters']}")
print(f"affinity: {best_params['affinity']}")
if best_params['affinity'] == 'nearest_neighbors':
    print(f"n_neighbors: {best_params['n_neighbors']}")
else:
    print(f"gamma: {best_params['gamma']}")
print(f"Silhouette score: {best_params['silhouette_score']:.3f}")

# Visualize the best clustering result
best_clusters = results_df.iloc[0]['cluster_labels']
df_cleaned['Cluster_Spectral_Tuned'] = best_clusters

# Visualize optimized clusters
plt.figure(figsize=(12, 8))
# Plot optimized clustering
plt.subplot(1, 2, 2)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=df_cleaned['Cluster_Spectral_Tuned'], palette='tab10', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title(f'Optimized Spectral Clustering\nSilhouette: {best_params["silhouette_score"]:.3f}')
plt.legend(title='Cluster')
print("\nData points in each optimized cluster:")
print(df_cleaned['Cluster_Spectral_Tuned'].value_counts())

plt.tight_layout()
plt.show()
# Create summary of the optimized clusters
cluster_summary_spectral_tuned = df_cleaned.groupby('Cluster_Spectral_Tuned').mean()
# Count the number of data points in each optimized cluster
```

## BIRCH(Tuned)

threshold: The maximum radius of the subcluster (in feature space) stored at each leaf of the CF tree. Larger values bring about lesser and larger subclusters.

branching_factor: Maximum number of CF subclusters in each node. If a new samples enters such that the number of subclusters exceed the branching_factor then that node is split into two nodes with the subclusters redistributed in each. The parent subcluster of that node is removed and two new subclusters are added as parents of the 2 split nodes. Larger values causes wider but shallower tree: Each node can hold many children before splitting.

n_clusters: number of clusters

https://scikit-learn.org/stable/modules/generated/sklearn.cluster.Birch.html

```python

# Define hyperparameter grid
param_grid = {
    'threshold': [0.1, 0.2, 0.3, 0.5],
    'branching_factor': [10, 30, 50],
    'n_clusters': [2,3,4,5]  # None lets BIRCH determine the number of clusters
}

# Create lists to store results
results = []

# Iterate through hyperparameters
print("Running hyperparameter tuning for BIRCH clustering...")
for threshold in tqdm(param_grid['threshold']):
    for branching_factor in param_grid['branching_factor']:
        for n_clusters in param_grid['n_clusters']:
            # Create model with current hyperparameters
            birch = Birch(
                threshold=threshold,
                branching_factor=branching_factor,
                n_clusters=n_clusters
            )
            
            # Fit model and predict clusters
            cluster_labels = birch.fit_predict(X_pca)
            
            # Compute silhouette score if there's more than one cluster
            if len(set(cluster_labels)) > 1:
                sil_score = silhouette_score(X_pca, cluster_labels)
                
                # Get actual number of clusters (important when n_clusters=None)
                n_clusters_actual = len(np.unique(cluster_labels))
                
                # Save results
                results.append({
                    'threshold': threshold,
                    'branching_factor': branching_factor,
                    'n_clusters_param': n_clusters,
                    'n_clusters_actual': n_clusters_actual,
                    'silhouette_score': sil_score,
                    'cluster_labels': cluster_labels
                })
            else:
                print(f"Skipping parameters (threshold={threshold}, branching_factor={branching_factor}, n_clusters={n_clusters}) - only one cluster formed")

# Convert results to DataFrame and sort by silhouette score
results_df = pd.DataFrame(results)
results_df = results_df.sort_values('silhouette_score', ascending=False)

# Get the best parameters
best_params = results_df.iloc[0]
print("\nBest parameters:")
print(f"Threshold: {best_params['threshold']}")
print(f"Branching factor: {best_params['branching_factor']}")
print(f"n_clusters parameter: {best_params['n_clusters_param']}")
print(f"Actual clusters formed: {best_params['n_clusters_actual']}")
print(f"Silhouette score: {best_params['silhouette_score']:.3f}")

# Apply the best model
best_birch = Birch(
    threshold=best_params['threshold'],
    branching_factor=best_params['branching_factor'],
    n_clusters=best_params['n_clusters_param']
)
df_cleaned['Cluster_BIRCH_Tuned'] = best_birch.fit_predict(X_pca)
print(df_cleaned['Cluster_BIRCH_Tuned'].value_counts())

# Visualize clusterings
plt.figure(figsize=(12, 8))
# Plot optimized clustering
plt.subplot(1, 2, 2)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=df_cleaned['Cluster_BIRCH_Tuned'], palette='tab10', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title(f'Optimized BIRCH Clustering\nSilhouette: {best_params["silhouette_score"]:.3f}')
plt.legend(title='Cluster')

plt.tight_layout()
plt.show()

cluster_summary_birch_tuned = df_cleaned.groupby('Cluster_BIRCH_Tuned').mean()


```
